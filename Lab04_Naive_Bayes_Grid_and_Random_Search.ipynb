```python
from sklearn.datasets import fetch_20newsgroups
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics import classification_report
from sklearn.model_selection import GridSearchCV, RandomizedSearchCV

from sklearn.naive_bayes import MultinomialNB
from sklearn.pipeline import Pipeline

from scipy.stats import uniform
```

```python
train = fetch_20newsgroups(subset='train')
test = fetch_20newsgroups(subset='test')

Xtrain = train.data[:3000]
ytrain = train.target[:3000]
Xtest = test.data[:500]
ytest = test.target[:500]

print("X:", len(Xtest))
print("y:", len(ytest))
```

### Naive Bayes [(Documentation)](https://scikit-learn.org/stable/modules/generated/sklearn.naive_bayes.MultinomialNB.html)

```python
pipeline =  Pipeline([('tfidf', TfidfVectorizer(stop_words = 'english')),
                      ('nb', MultinomialNB())])
pipeline.fit(Xtrain,ytrain)
ypred = pipeline.predict(Xtest)

print(classification_report(ytest, ypred))
```

### Random Search Cross-Validation [(Documentation)](https://scikit-learn.org/stable/modules/generated/sklearn.model_selection.RandomizedSearchCV.html)

### Uniform distribution in `Scipy` [(Documentation)](https://docs.scipy.org/doc/scipy/reference/generated/scipy.stats.uniform.html)

```python
parameters = {'nb__alpha': uniform(loc =0,scale = 10)}

clf = RandomizedSearchCV(pipeline, parameters, n_iter=7, verbose=100)
clf.fit(Xtrain, ytrain)
```

```python
ypred = clf.predict(Xtest)

print(classification_report(ytest, ypred))
```

# Gridsearch


```python
import numpy as np
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import f1_score


params = {'alpha__alpha': np.logspace(-3, 1, 10)}

gridcv = GridSearchCV(pipeline, params,
                      scoring='f1_macro', cv=5)

gridcv.fit(Xtrain, ytrain)

ypred_grid = gridcv.predict(Xtest)
print(classification_report(ytest, ypred_grid))
```

```python
from sklearn.metrics import f1_score

f1_macro_grid = f1_score(ytest, ypred_grid, average='macro')

print("Best alpha:", gridcv.best_params_['alpha__alpha'])
print(f"f1_macro score on test set: {f1_macro_grid:.4f}")
```

# Random search

```python
params ={'alpha__alpha': np.logspace(-1,3,10)}

random_search_cv = RandomizedSearchCV(pipeline,
                                     params,
                                     n_iter=10,
                                     scoring='f1_macro',
                                     cv=5,
                                     verbose=100,)

random_search_cv.fit(Xtrain, ytrain)
```

```python
ypred_random = random_search_cv.predict(Xtest)

print(classification_report(ytest, ypred_random))
```

```python
from sklearn.metrics import f1_score

f1_macro_grid = f1_score(ytest, ypred_random, average='macro')

print("Best alpha:", random_search_cv.best_params_['alpha__alpha'])
print(f"f1_macro score on test set: {f1_macro_grid:.4f}")
```
